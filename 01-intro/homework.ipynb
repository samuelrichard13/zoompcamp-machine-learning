```python
horsepower_frequency = df.horsepower.value_counts()
horsepower_frequency
```

```python
most_frequent_hp = horsepower_frequency.index[horsepower_frequency == horsepower_frequency.max()][0]
most_frequent_hp
```

```python
df_filled = df
df_filled['horsepower'] = df['horsepower'].fillna(most_frequent_hp)
```

```python
df_asia = df[df.origin == 'Asia']
df_asia[['vehicle_weight', 'model_year']].head(7)
```

```python
import numpy as np
```

```python
X = df_asia[['vehicle_weight', 'model_year']].head(7).values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
XTX_inv.dot(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```
